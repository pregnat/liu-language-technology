```python
text = """Fredrik Backman is a Swedish author, blogger, and columnist. He grew up in Helsingborg,
Sweden and debuted as a novelist in 2012 with A Man Called Ove. The novel was published in English
in 2013 and reached the New York Times Best Seller list 18 months after its publication. Backman's
books have been published in more than twenty-five languages. Fredrik Backman married Neda Shafti
Backman in 2009.""".replace("\n", " ")
```

To analyze this paragraph with the help of spaCy, we need to import spacy and load a “model”.  The following cell loads the model called `en_core_web_md`, a medium-sized English pipeline trained on web data – you could [check the spaCy model list](https://spacy.io/usage/models) to see what other models are available.
```python
import spacy

nlp = spacy.load("en_core_web_md")
```

Now we just need to call this model on our example text to get a spaCy document:

```python
doc = nlp(text)
doc
```

The spaCy document looks just like normal text, but it is actually a Python object with lots of attributes and functions that we can use to analyze the text.  For example, spaCy automatically performs **sentence segmentation**, i.e. it splits the text into individual sentences.  We can access them with the `.sents` attribute.  The following cell prints each sentence on a separate line:

```python
for sentence in doc.sents:
    print(sentence.text)
```

spaCy also automatically performs **tokenization**, which divides the text into even smaller units called **tokens**.  We will learn more about tokenization throughout the course, as it is an important part of every language technology pipeline.  In spaCy, we can access the individual tokens by simply iterating over the document (or a sentence), like this:

```python
for token in doc:
    print(token.text)
```

## Task 2: Extracting part-of-speech tags with spaCy

Of course, spaCy can do more than just segment the text into smaller units.  Each _token_ in spaCy has a number of attributes corresponding to different _features_ of the token, such as:

- `.text` contains the token as a simple Python string.
- `.lemma_` contains the _lemma_ of the token – we will learn more about what a “lemma” is later in the course.
- `.is_alpha` is True if the token contains any alphabetic characters, and False if it e.g. only contains punctuation marks or digits.
- `.pos_` contains the _part of speech (POS)_ tag of the token – this tells us whether the token is, for example, a *noun*, *verb*, *adjective*, *pronoun*, etc.
- `.morph` contains _morphological features_ of the token, such as whether a pronoun is *masculine* or *feminine*, or whether a verb is in *past* or *present tense*.

The following cell demonstrates how to access this information, and displays it in a nice way (in form of a table). You don't need to fully understand what all of this means yet, but if you want to learn more, the [spaCy documentation on “linguistic features”](https://spacy.io/usage/linguistic-features) explains all of this in detail.

```python
# Let's take the second sentence from the document
sentence = list(doc.sents)[1]

# We produce a list of each token, together with its part-of-speech label
tokens = []
for token in sentence:
    tokens.append([token.text, token.lemma_, token.is_alpha, token.pos_, token.morph])

# Display the list in a nice way
tabulate(tokens, tablefmt="html", headers=["Token", "Lemma", "Alpha", "POS", "Morphology"])
```

There is even a function to “explain” what certain POS tags mean – for example, if you want to know what "PROPN" stands for, you can do:

```python
spacy.explain("PROPN")
```

<div class="alert alert-info">

  **Complete the function below to extract a list of all verbs from a spaCy document!**
  
  - _Note:_ Assume that “verbs” are all tokens that are tagged with **either** “VERB” **or** “AUX” (auxiliary verb)!

</div>

```python
def extract_verbs(doc) -> list[str]:    

    verbs = []
    for token in doc:
        if token.pos_ == "VERB" or token.pos_ == "AUX":
            verbs.append(token.text)
    return verbs
```

## Task 3: Extracting named entities with spaCy

Another feature of spaCy is that it can extract **named entities**.  We will learn more about what “named entities” are later in the course, but since spaCy makes it so easy to access them, let’s just take a look at what spaCy gives us when we access the `.ents` attribute of a document:

```python
doc.ents
```

You can see that this returns a list containing, among other things, names of people ("Fredrik Backman", "Backman", "Neda Shafti Backman") and places ("Helsingborg", "Sweden").  However, just as with “sentences” or “tokens”, they are not just strings, but have attributes with more information.  spaCy provides a nice way to visualize these “entities”:

```python
from spacy import displacy

displacy.render(doc, style="ent")
```

Even without knowing anything about “named entity recognition”, you can see that spaCy highlights parts of the text as referring to a “person”, “language”, or “work of art”.  Some of the labels are not so self-explanatory, and you might be wondering what “NORP” or “GPE” stands for.  Let’s use `spacy.explain` to check:

```python
print("NORP stands for: ", spacy.explain("NORP"))
print(" GPE stands for: ", spacy.explain("GPE"))
```
